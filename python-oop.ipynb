```python
class Person:
    species = 'Homo Sapiens'
    
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def myfunc(self):
        print('Hi, I am ' + self.name)

    @classmethod
    def from_birth_year(cls, name, birth_year, current_year=2026):
        age = current_year - birth_year
        return cls(name, age)

    @staticmethod
    def is_adult(age):
        return age >= 10

person_one = Person('Hannan', 22)
person_one.myfunc()

person_two = Person.from_birth_year('Hannan', 2003)
print(person_two.name, person_two.age)

print('Does person one meet the adult age check?', Person.is_adult(person_one.age))
```

Inheritance

```python
class Person:
    def __init__(self, given_name, family_name):
        self.firstname = given_name
        self.lastname = family_name

    def printname(self):
        print(self.firstname, self.lastname)

example_value = Person('Hannan', 'Shahid')
example_value.printname()
```

```python
class Student(Person):
    pass

example_value = Student('Hannan', 'Shahid')
example_value.printname()
```

```python
class Student(Person):
    def __init__(self, given_name, family_name, student_id):
        super().__init__(given_name, family_name)
        self.student_id = student_id

student_record = Student('Hannan', 'Shahid', 13)
student_record.printname()
print(student_record.student_id)
```

```python
class Truck:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def move(self):
        print('Drive along the road')

class Ferry:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def move(self):
        print('Travel across the water')

class Aircraft:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def move(self):
        print('Fly through the air')

truck_example = Truck('Volvo', 'FH')
ferry_example = Ferry('Damen', 'Fast Ferry')
aircraft_example = Aircraft('Airbus', 'A320')

for transport_item in (truck_example, ferry_example, aircraft_example):
    transport_item.move()
```

```python
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def move(self):
        print('Travel forward')

class Truck(Vehicle):
    pass

class Ferry(Vehicle):
    def move(self):
        print('Travel across the water')

class Aircraft(Vehicle):
    def move(self):
        print('Fly through the air')

truck_example = Truck('Volvo', 'FH')
ferry_example = Ferry('Damen', 'Fast Ferry')
aircraft_example = Aircraft('Airbus', 'A320')

for transport_item in (truck_example, ferry_example, aircraft_example):
    print(transport_item.brand, transport_item.model)
    transport_item.move()
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name #public
        self._dept = 'Engineering' #protected
        self.__salary = salary #private

    def get_salary(self):
     return self.__salary

    def raise_salary(self, percentage):
            self.__salary *= (1 + percentage / 100)
            return self.__salary

staff_member = Employee('Sara', 100000)
print(staff_member.name)
print(staff_member._dept)
print(staff_member.get_salary())

try:
    print(staff_member.__salary)
except AttributeError as access_error:
    print('Private attribute access failed:', staff_member)
```
