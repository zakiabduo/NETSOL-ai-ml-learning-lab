```python
#Data Abstraction
#Abstract Base Class.
from abc import ABC, abstractmethod
class Person(ABC):
    def __init__(self, full_name, member_id):
        self.full_name = full_name              # Public
        self._member_id = member_id   # Protected
        self.__identity_number = "Hidden"        # Private
    @abstractmethod
#abstractmethod → marks a method that subclasses must implement.
    def display_role(self):
        pass
    def get_cnic(self):
        return self.__identity_number
```

```python
#3. Single Inheritance
from abc import ABC, abstractmethod
class Person(ABC):
    def __init__(self, full_name, member_id):
        self.full_name = full_name              # Public
        self._member_id = member_id   # Protected
        self.__identity_number = "Hidden"        # Private
    @abstractmethod
    def display_role(self):
        pass
    def get_cnic(self):
        return self.__identity_number
```

```python
#4. Employee → Faculty
class Employee(Person):
    def __init__(self, full_name, member_id, pay_amount):
        super().__init__(full_name, member_id)
        self.__pay_amount = pay_amount
    def display_role(self):
        print(f"{self.full_name} works as an employee.")
    def get_salary(self):
        return self.__pay_amount
    def set_salary(self, pay_amount):
        if pay_amount > 0:
            self.__pay_amount = pay_amount
        else:
            print("Salary must be greater than zero.")
```

```python
#5. Multilevel Inheritance
#Person
#   ↓
#Employee
#   ↓
#Faculty
#   ↓
#Professor
class Faculty(Employee):
    def __init__(self, full_name, member_id, pay_amount, academic_department):
        super().__init__(full_name, member_id, pay_amount)
        self.academic_department = academic_department
    def display_role(self):
        print(f"{self.full_name} belongs to the faculty of {self.academic_department}.")
```

```python
class Professor(Faculty):
    def __init__(self, full_name, member_id, pay_amount, academic_department, expertise):
        super().__init__(
            full_name,
            member_id,
            pay_amount,
            academic_department
        )
        self.expertise = expertise
    def display_role(self):
        print(
            f"{self.full_name} works as a professor "
            f"with expertise in {self.expertise}."
        )
    def conduct_research(self):
        print(
            f"{self.full_name} is carrying out research "
            f"in {self.expertise}."
        )
```

```python
professor_example = Professor(
    "Dr. Kamran",
    "F001",
    200000,
    "Computer Science",
    "Artificial Intelligence"
)
professor_example.display_role()
professor_example.conduct_research()
```

```python
#6. Hierarchical Inheritance
#             Faculty
#             /     \
#            /       \
#     Professor     Lecturer
class Lecturer(Faculty):
    def __init__(self, full_name, member_id, pay_amount, academic_department, teaching_courses):
        super().__init__(
            full_name,
            member_id,
            pay_amount,
            academic_department
        )
        self.teaching_courses = teaching_courses
    def display_role(self):
        print(f"{self.full_name} works as a lecturer.")
    def teach(self):
        print(
            f"{self.full_name} teaches the following courses: "
            f"{', '.join(self.teaching_courses)}"
        )
```

```python
lecturer_example = Lecturer(
    "Ms. Nadia",
    "L001",
    120000,
    "Software Engineering",
    ["Python", "OOP"]
)

lecturer_example.display_role()
lecturer_example.teach()
```

```python
#7. Multiple Inheritance
class Researcher:
    def __init__(self, research_topic):
        self.research_topic = research_topic

    def conduct_research(self):
        print(
            f"Research focus: {self.research_topic}."
        )
```

```python
class ResearchProfessor(Professor, Researcher):
    def __init__(
        self,
        full_name,
        member_id,
        pay_amount,
        academic_department,
        expertise,
        research_topic
    ):
        Professor.__init__(
            self,
            full_name,
            member_id,
            pay_amount,
            academic_department,
            expertise
        )
        Researcher.__init__(
            self,
            research_topic
        )
```

```python
research_faculty_example = ResearchProfessor(
    "Dr. Imran",
    "R001",
    250000,
    "Artificial Intelligence",
    "Machine Learning",
    "Explainable AI"
)
research_faculty_example.display_role()
research_faculty_example.conduct_research()
```

```python
from abc import ABC, abstractmethod
# ==========================================
# ABSTRACT BASE CLASS
# ==========================================
class Person(ABC):
    def __init__(self, full_name, member_id):
        self.full_name = full_name              # Public
        self._member_id = member_id   # Protected
        self.__identity_number = "Hidden"        # Private
    @abstractmethod
    def display_role(self):
        pass
    def get_cnic(self):
        return self.__identity_number
# ==========================================
# SINGLE INHERITANCE
# Person → Student
# ==========================================
class Student(Person):
    def __init__(self, full_name, member_id, degree_program):
        super().__init__(full_name, member_id)
        self.degree_program = degree_program
    def display_role(self):
        print(f"{self.full_name} is registered as a student.")
    def enroll_course(self, course_title):
        print(f"{self.full_name} has has enrolled in {course_title}.")
# ==========================================
# Person → Employee
# ==========================================
class Employee(Person):
    def __init__(self, full_name, member_id, pay_amount):
        super().__init__(full_name, member_id)
        self.__pay_amount = pay_amount
    def display_role(self):
        print(f"{self.full_name} works as an employee.")
    # Getter
    def get_salary(self):
        return self.__pay_amount
    # Setter
    def set_salary(self, pay_amount):
        if pay_amount > 0:
            self.__pay_amount = pay_amount
        else:
            print("Salary must be greater than zero.")
# ==========================================
# MULTILEVEL INHERITANCE
# Person → Employee → Faculty → Professor
# ==========================================
class Faculty(Employee):
    def __init__(self, full_name, member_id, pay_amount, academic_department):
        super().__init__(
            full_name,
            member_id,
            pay_amount
        )
        self.academic_department = academic_department

    def display_role(self):
        print(
            f"{self.full_name} is a faculty member "
            f"of {self.academic_department}."
        )
class Professor(Faculty):
    def __init__(
        self,
        full_name,
        member_id,
        pay_amount,
        academic_department,
        expertise
    ):
        super().__init__(
            full_name,
            member_id,
            pay_amount,
            academic_department
        )
        self.expertise = expertise
    def display_role(self):
        print(
            f"{self.full_name} works as a professor "
            f"with expertise in {self.expertise}."
        )
    def conduct_research(self):
        print(
            f"{self.full_name} is carrying out research "
            f"in {self.expertise}."
        )
# ==========================================
# HIERARCHICAL INHERITANCE
#
#                 Faculty
#                /       \
#          Professor    Lecturer
# ==========================================
class Lecturer(Faculty):
    def __init__(
        self,
        full_name,
        member_id,
        pay_amount,
        academic_department,
        teaching_courses
    ):
        super().__init__(
            full_name,
            member_id,
            pay_amount,
            academic_department
        )
        self.teaching_courses = teaching_courses
    def display_role(self):
        print(f"{self.full_name} works as a lecturer.")
    def teach(self):
        print(
            f"{self.full_name} teaches the following courses: "
            f"{', '.join(self.teaching_courses)}"
        )
# ==========================================
# MULTIPLE INHERITANCE
# ==========================================
class Researcher:
    def __init__(self, research_topic):
        self.research_topic = research_topic
    def conduct_research(self):
        print(
            f"Research focus: {self.research_topic}."
        )
class ResearchProfessor(Professor, Researcher):
    def __init__(
        self,
        full_name,
        member_id,
        pay_amount,
        academic_department,
        expertise,
        research_topic
    ):
        Professor.__init__(
            self,
            full_name,
            member_id,
            pay_amount,
            academic_department,
            expertise
        )
        Researcher.__init__(
            self,
            research_topic
        )
# ==========================================
# CREATE OBJECTS
# ==========================================
print("\n--- STUDENT DETAILS ---")
student_example = Student(
    "Zain",
    "ST001",
    "BS Software Engineering"
)
student_example.display_role()
student_example.enroll_course("Object Oriented Programming")
print("\n--- LECTURER DETAILS ---")
lecturer_example = Lecturer(
    "Nadia",
    "L001",
    120000,
    "Software Engineering",
    ["Python", "OOP"]
)
lecturer_example.display_role()
lecturer_example.teach()
print("\n--- PROFESSOR DETAILS ---")
professor_record = Professor(
    "Dr. Kamran",
    "P001",
    200000,
    "Computer Science",
    "Artificial Intelligence"
)
professor_record.display_role()
professor_record.conduct_research()
print("\n--- RESEARCH PROFESSOR DETAILS ---")
research_professor_record = ResearchProfessor(
    "Dr. Imran",
    "RP001",
    250000,
    "Artificial Intelligence",
    "Machine Learning",
    "Explainable AI"
)
research_professor_record.display_role()
research_professor_record.conduct_research()
print("\n--- ATTRIBUTE ACCESS AND SALARY ---")
print("Name (public attribute):", student_example.full_name)
print("Member ID (protected attribute):", student_example._member_id)
print("Private identity value via getter:",
      student_example.get_cnic())
print(
    "Current professor salary:",
    professor_record.get_salary()
)
professor_record.set_salary(220000)
print(
    "Salary after update:",
    professor_record.get_salary()
)
```

```python
#Inner / Nested Classes
class Computer:
    def __init__(self, manufacturer):
        self.manufacturer = manufacturer
        self.processor = self.CPU("Intel i7", 8)     # build an inner object

    class CPU:                                  # nested class
        def __init__(self, processor_model, core_count):
            self.processor_model = processor_model
            self.core_count = core_count

        def spec(self):
            return f"{self.processor_model} ({self.core_count} processing cores)"

    def describe(self):
        return f"{self.manufacturer} equipped with {self.processor.spec()}"


computer_example = Computer("Dell")
print(computer_example.describe())

# The inner class can also be used on its own:
processor_example = Computer.CPU("AMD Ryzen 9", 16)
print(processor_example.spec())
```

```python
class BankAccount:
    def __init__(self, account_holder, account_balance=0):
        self.account_holder = account_holder          # instance variable
        self.__account_balance = account_balance     # private attribute
    def deposit(self, transaction_amount):      # instance method
        self.__account_balance += transaction_amount
        print("Account balance:", self.__account_balance)
        return self.__account_balance


    def withdraw(self, transaction_amount):
        if transaction_amount > self.__account_balance:
            print("Account balance:", self.__account_balance)
            print("Requested withdrawal:", transaction_amount)
            return "The withdrawal exceeds the available balance"

        self.__account_balance -= transaction_amount
        return self.__account_balance
        print(self.__account_balance)

primary_account = BankAccount("Mariam", 1000)
secondary_account = BankAccount("Omar")
primary_account.deposit(500)
secondary_account.deposit(200)
#print(a.owner, a.__balance)
#print(b.owner, b.__balance)
print(primary_account.withdraw(99999))
# The two calls below are equivalent:
print(primary_account.deposit(10), BankAccount.deposit(primary_account, 10))
```

```python
class BankAccount:
    def __init__(self, account_holder, account_balance=0):
        self.account_holder = account_holder
        self.__account_balance = account_balance     # private attribute

    @property
    def account_balance(self):
        return self.__account_balance

    def deposit(self, transaction_amount):
        if transaction_amount <= 0:
            raise ValueError("A deposit must be greater than zero")
        self.__account_balance += transaction_amount

    def withdraw(self, transaction_amount):
        if transaction_amount > self.__account_balance:
            raise ValueError("The withdrawal exceeds the available balance")
        self.__account_balance -= transaction_amount

account_example = BankAccount("Nadia", 100)
account_example.deposit(50)
account_example.withdraw(30)
print("Available account balance:", account_example.account_balance)     # read like a normal attribute, but it's protected internally

try:
    account_example.withdraw(10000)
except ValueError as transaction_error:
    print("Transaction failed:", transaction_error)
```
