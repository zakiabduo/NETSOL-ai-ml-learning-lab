```python
prod_dict = {'headphones':15, 'keyboard':20, 'laptop':30, 'monitor':50, 'webcam':3}
print(prod_dict)

print(prod_dict['headphones'])

prod_dict['charger'] = 3
print(prod_dict)
```

```python
input_devices = {'mouse', 'keyboard'}
output_devices = {'monitor', 'headphones'}
computers = {'laptop'}
print(input_devices)
print(output_devices)
print(computers)
```

```python
while True:
    print("\n=====INVENTORY MANAGEMENT SYSTEM====")
    print("1. Display all products")
    print("2. Add a product")
    print("3. Remove a product")
    print("4. Check stock")
    print("5. Update stock")
    print("6. Sell product")
    print("7. Display product categories")
    print("8. Exit")
    
    choice = input("Enter your choice: ")
    
    if choice == '1':
        print(f"Products in inventory: {products_list}")
        print(f"Total number of products: {len(products_list)}")
        
    elif choice == '2':
        item = input("Add product: ")
    
        if item in products_list:
            print("Product already exists.")
        else:
            products_list.append(item)
            prod_dict[item] = 0
            print("Added successfully!")
            
    elif choice == '3':
        item = input("Remove product: ")
    
        if item in products_list:
            products_list.remove(item)
            del prod_dict[item]
            print("Removed successfully!")
        else:
            print("Product not found.")
            
    elif choice == '4':
        item = input("Check product: ")
    
        if item in prod_dict:
            print("Stock:", prod_dict[item])
        else:
            print("Product not found.")
        
    elif choice == '5':
        item = input("Enter product: ")
        if item in prod_dict:
            new_qty = int(input("Enter new quantity: "))
            prod_dict[item] = new_qty
            print("Stock updated successfully!")
        else:
            print("Product not found. Please add it first.")
            
    elif choice == '6':
        item = input("Sell product: ")
    
        if item in prod_dict and prod_dict[item] > 0:
            prod_dict[item] -= 1
            print("Sold! Remaining:", prod_dict[item])
        else:
            print("Not found or out of stock.")
            
    elif choice == '7':
        print(f"Input Devices: {input_devices}")
        print(f"Output Devices: {output_devices}")
        print(f"Computers: {computers}")
        
    elif choice == '8':
        print("Exiting Inventory System.")
        break 
        
    else:
        print("Invalid choice. Please enter a number from 1 to 8.")
```
