```python
# Import the libraries and display plots inside the notebook.
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from sklearn import datasets, linear_model
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Load the diabetes dataset for the regression example.
diabetes = datasets.load_diabetes()
```

```python
# Select BMI as the single input feature and keep it two-dimensional.
X = diabetes.data[:, np.newaxis, 2]
```

```python
# Use the last 30 feature rows for testing and the remaining rows for training.
X_train = X[:-30]
X_test = X[-30:]
```

```python
# Split the target values using the same training and test rows.
y_train = diabetes.target[:-30]
y_test = diabetes.target[-30:]
```

```python
# Create a linear regression model.
regression_model = linear_model.LinearRegression()
```

```python
# Fit the model using the training features and target values.
regression_model.fit(X_train, y_train)
```

```python
# Predict disease progression scores for the test data.
y_pred = regression_model.predict(X_test)
```

```python
# Print the model coefficient, mean squared error, and R-squared score.
print('Coefficients: \n', regression_model.coef_)
print("Mean squared error: %.2f" % mean_squared_error(y_test, y_pred))
print('Variance score: %.2f' % r2_score(y_test, y_pred))
```

```python
# Plot the test data and the predicted regression line.
plt.scatter(X_test, y_test, color = 'blue')
plt.plot(X_test, y_pred, color = 'red', linewidth = 3)
plt.xticks(())
plt.yticks(())
plt.show()
```
