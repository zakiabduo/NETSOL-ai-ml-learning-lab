```python
import time

item_count = 1_000_000
list_values = list(range(item_count))
array_values = np.arange(item_count)

start_time = time.time()
doubled_list = [item * 2 for item in list_values]
list_duration = time.time() - start_time

start_time_2 = time.time()
doubled_array = array_values * 2
array_duration = time.time() - start_time_2

print(f"Time using a Python loop: {list_duration:.4f} seconds")
print(f"Time using NumPy vectorization: {array_duration:.4f} seconds")
print(f"Approximate NumPy speedup: {list_duration / array_duration:.1f}x")
```

```python
import os

os.makedirs("sample_data", exist_ok=True)

score_values_7 = np.array([78, 85, 91, 67, 88])

np.save("sample_data/marks.npy", score_values_7)
restored_scores = np.load("sample_data/marks.npy")
print(restored_scores)
print(np.array_equal(score_values_7, restored_scores))
```
