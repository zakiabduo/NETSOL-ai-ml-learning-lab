### Task 9: Combining Arrays
Create two 1D arrays, `group_a = np.array([1, 2, 3])` and `group_b = np.array([4, 5, 6])`. Use `np.concatenate` to join them, and separately use `np.vstack` to stack them as two rows of a 2D array. Print both results.

### Task 10: Sorting and Uniqueness
Given `grades = np.array(["B", "A", "C", "A", "B", "A", "D"])`, print the unique grades along with a count of how many times each occurs, sorted so the most common grade appears first (hint: use `return_counts=True`, then sort by the counts).

````python
# Task 6 -- Your code here
scores = np.array([55, 89, 76, 92, 40, 67, 98])
greater_70 = scores[scores >= 70]
print('Scores greater than 70: ', greater_70)
print('Total number of scores greater than 70: ', len(greater_70))
````

````python
# Task 8 -- Your code here
original = np.array([5, 10, 15, 20, 25])
partial_view = original[1:3]
print('Original')
print(original)
print(partial_view, '\n')

print('Edited')
partial_view[0] = 123
print(original)
print(partial_view, '\n')

print('Copied')
partial_view = original[1:3].copy()
partial_view[0] = 321
print(original)
print(partial_view)
````

````python
# Task 10 -- Your code here
grades = np.array(["B", "A", "C", "A", "B", "A", "D"])

unique_grades, counts = np.unique(grades, return_counts=True)
print("Unsorted:", unique_grades, counts)


order = np.argsort(counts)[::-1]


print("Sorted Grades:", unique_grades[order])
print("Sorted Counts:", counts[order])
````

### Task 11: Normalizing a Feature Column
You are given a raw feature column representing house sizes in square feet:
```python
house_sizes = np.array([850, 900, 1200, 1500, 2000, 750, 1100])
```
Normalize this array so every value is scaled into the range `[0, 1]`, using the formula `(value - min) / (max - min)`. Do this in a single vectorized line (no loop), then print the result.

### Task 12: A Toy Prediction with Matrix Multiplication
You are given a small feature matrix (3 samples, 3 features each) and a fixed weight vector:
```python
X = np.array([
    [2, 3, 1],
    [5, 1, 4],
    [1, 1, 1],
])
weights = np.array([0.4, 0.4, 0.2])
```
Compute the predictions for all three samples in a single line using `@`. Then find the index of the sample with the **highest** predicted value using `np.argmax`.

### Task 13: Per-student Analysis with `axis`
You are given marks for 4 students across 3 subjects:
```python
student_marks = np.array([
    [78, 85, 90],
    [88, 92, 95],
    [55, 62, 58],
    [90, 87, 93],
])
```
Using only vectorized NumPy operations (no Python loops):
1. Compute each student's average mark (`axis=1`).
2. Compute the class average for each subject (`axis=0`).
3. Use boolean indexing to find which students have an average `>= 80`.
4. Save the array of per-student averages to a file `student_averages.npy`, then load it back and print it.

````python
# Task 11 -- Your code here
house_sizes = np.array([850, 900, 1200, 1500, 2000, 750, 1100])

normalized_sizes = (house_sizes - house_sizes.min()) / (house_sizes.max() - house_sizes.min())
print(normalized_sizes)
````

````python
# Task 13 -- Your code here
student_marks = np.array([
    [78, 85, 90],
    [88, 92, 95],
    [55, 62, 58],
    [90, 87, 93],
])

student_avgs = student_marks.mean(axis=1)
print('Student Averages:', student_avgs)

subject_avgs = student_marks.mean(axis=0)
print('Subject Averages:', subject_avgs)

top_student_avgs = student_avgs[student_avgs >= 80]
print('Averages >= 80:', top_student_avgs)

np.save('student_averages.npy', student_avgs)
loaded_avgs = np.load('student_averages.npy')
print('Loaded Averages:', loaded_avgs)
````

````python
# Task 12 -- Your code here
X = np.array([
    [2, 3, 1],
    [5, 1, 4],
    [1, 1, 1],
])
weights = np.array([0.4, 0.4, 0.2])

predictions = X @ weights
print("Predictions:", predictions)

highest_index = np.argmax(predictions)
print("Index of highest prediction:", highest_index)

````
