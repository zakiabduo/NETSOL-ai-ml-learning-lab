```python
# Import Pandas and NumPy, then print the Pandas version.
import pandas as pd
import numpy as np

print(pd.__version__)
```

```python
# Create a Series of scores and inspect its values and type.
score_series = pd.Series([78, 85, 91, 67, 88])
print(score_series)
print(type(score_series))
```

```python
# Create labeled subject scores, select Science, and calculate the mean.
subject_scores = pd.Series([78, 85, 91], index=["Math", "Science", "English"])
print(subject_scores)

print(subject_scores["Science"])
print(subject_scores.mean())
```

```python
# Build a student DataFrame from a dictionary of columns.
student_columns = {
    "name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "age": [22, 21, 23, 20],
    "marks": [84, 91, 58, 90],
}

student_table = pd.DataFrame(student_columns)
print(student_table)
print(type(student_table))
```

```python
# Create a DataFrame from dictionaries representing individual students.
student_records = [
    {"name": "Ali", "age": 22, "marks": 84},
    {"name": "Sara", "age": 21, "marks": 91},
    {"name": "Ahmed", "age": 23, "marks": 58},
]

records_table = pd.DataFrame(student_records)
print(records_table)
```

```python
# Convert a NumPy score matrix into a DataFrame with subject columns.
score_matrix = np.array([
    [78, 85, 90],
    [88, 92, 95],
])

subject_table = pd.DataFrame(score_matrix, columns=["Math", "Science", "English"])
print(subject_table)
```

```python
# Filter scores of at least 80 using a boolean mask or a direct condition.
student_table = pd.read_csv("sample_data/students.csv")

high_score_filter = student_table["marks"] >= 80
print(high_score_filter)

print(student_table[high_score_filter])


print(student_table[student_table["marks"] >= 80])
```

```python
# Combine two conditions to select Karachi students scoring at least 70.
print(student_table[(student_table["marks"] >= 70) & (student_table["city"] == "Karachi")])
```

```python
# Add a passed column using a minimum score of 60.
student_table = pd.read_csv("sample_data/students.csv")

student_table["passed"] = student_table["marks"] >= 60
print(student_table)
```

```python
# Increase every student score by five marks.
student_table["marks"] = student_table["marks"] + 5
print(student_table)
```

```python
# Drop the city column into a new table and retain the original columns.
student_table = pd.read_csv("sample_data/students.csv")

table_without_city = student_table.drop(columns=["city"])
print(table_without_city)
print()
print(student_table.columns)
```

```python
# Create a new table without the row labeled zero.
table_without_first_row = student_table.drop(index=0)
print(table_without_first_row)
```

```python
# Build a student table containing missing age and score values.
incomplete_records = {
    "name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "age": [22, np.nan, 23, 20],
    "marks": [84, 91, np.nan, 90],
}

student_table = pd.DataFrame(incomplete_records)
print(student_table)
```

```python
# Locate missing entries and count them in each column.
print(student_table.isna())
print()
print(student_table.isna().sum())
```

```python
# Keep only rows that contain no missing values.
complete_rows = student_table.dropna()
print(complete_rows)
```

```python
# Fill missing ages and scores with their respective column means.
filled_table = student_table.fillna({"age": student_table["age"].mean(), "marks": student_table["marks"].mean()})
print(filled_table)
```

```python
# Sort scores in both directions, then sort by city and descending score.
student_table = pd.read_csv("sample_data/students.csv")

print(student_table.sort_values("marks"))
print()
print(student_table.sort_values("marks", ascending=False))
print()
print(student_table.sort_values(["city", "marks"], ascending=[True, False]))
```
